This function reads an element and prints if it is even or odd

```python
def fn_evven_odd(ele):
  if ele %2 ==0:
    print(ele, " - is even")
  else:
    print(ele, " - is odd")
```

```python
fn_evven_odd(ele=5)
```

```python
a = [1,2,3,4,5,6]
for ele in a:
  fn_evven_odd(ele)
```
